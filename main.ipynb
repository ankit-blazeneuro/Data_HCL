## learning about number system 

Number System 
decimal       10      No      prefix 25
binary        2       0b      0b11001
octal         8       0o      0o31
Hexadecimal   16      0x      0x19

```python
decimal_value = 10
binary_value = 0b1010
octal_value = 0o12
hexadecimal_value = 0xA

print(decimal_value, type(decimal_value))
print(binary_value, type(binary_value))
print(octal_value, type(octal_value))
print(hexadecimal_value, type(hexadecimal_value))
```

```python
def add(*num): # single asterisk is used to pass variable number of arguments to a function. It will take all the arguments and store them in a tuple.
    sum =0 
    for n in num:
        sum += n
    return sum

print(add())
print(add(1))
print(add(1, 2, 3, 4, 5))
```

```python
class student:
    def set_setails(self, name, roll_number, course):
        self.name = name
        self.roll_number = roll_number
        self.course = course

    def display_details(self):
        print(f"Student Name: {self.name}")
        print(f"Roll Number: {self.roll_number}")
        print(f"Course: {self.course}")

student1 = student()
student1.set_setails("Ankit", 101, "Python Programming")
student1.display_details()
```

# Polymorphism

```python
class SmartPhone:
    def __init__(self, brand, model):
        self.brand = brand 
        self.model = model 
        self.__battery = 100

    def use_app(self, minutes):
        self.__battery -= minutes * 0.5

        if self.__battery < 0:
            self.battery = 0

        print(f"Used {minutes} minutes of app. Battery level: {self.__battery}%")

    def charge(self):
        self.__battery = 100
        print("Phone charged. Battery level: 100%")

    def status(self):
        print(f"Brand: {self.brand}, Model: {self.model}, Battery: {self.__battery}%")

phone1 = SmartPhone("Samsung", "A35")

phone1.use_app(10)
phone1.charge()
phone1.status()
```

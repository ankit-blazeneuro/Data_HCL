## learning about number system 

Number System 
decimal       10      No      prefix 25
binary        2       0b      0b11001
octal         8       0o      0o31
Hexadecimal   16      0x      0x19

```python
decimal_value = 10
binary_value = 0b1010
octal_value = 0o12
hexadecimal_value = 0xA

print(decimal_value, type(decimal_value))
print(binary_value, type(binary_value))
print(octal_value, type(octal_value))
print(hexadecimal_value, type(hexadecimal_value))
```

## Learning About the number system conversion

```python
number = 5

print(bin(number), oct(number), hex(number))
```

```python
print(int("11001", 2))  #binary to decimal
print(int("31", 8))     #octal to decimal
print(int("1F", 16))    #hexadecimal to decimal 
```

```python
large_number = 2.5e6  #2.5*10^(6)
small_number = 3.2e-4 #3.2*10^(-4)

print(large_number, type(large_number))
print(small_number, type(small_number))
```

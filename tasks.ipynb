## Task 1

```python
# each order contain item_name, units, price, total_price, cancel

orders=[
    ["Mouse", 2, 800, 1600, True],
    ["Keyboard", 3, 1200, 3600, False],
    ["Monitor", 4, 15000, 60000, True],
    ["CPU", 5, 25000, 125000, False],
    ["Monitor", 6, 20000, 120000, True],
    ["GPU", 7, 30000, 210000, False]
]

for order in orders:
    print(f"Item: {order[0]}, Units: {order[1]}, Price: {order[2]}, Total Price: {order[3]}, Cancelled: {order[4]}")


```

## Task 2

```python
# calculating GST of total item 

for order in orders:
    gst = order[3] * 0.18  # Assuming GST is 18% of total price
    print(f"Item: {order[0]}, Total Price: {order[3]}, GST: {gst}")

# adding all the total price of all items

total_price_sum = sum(order[3] for order in orders)
print(f"Total Price of all items: {total_price_sum}")

```

## Task 3

```python
if total_price_sum >= 10000:
    print("Total price exceeds 10,000. Applying discount of 10%.")
    discount = total_price_sum * 0.10
    total_price_sum -= discount
    print(f"Discount: {discount}, New Total Price: {total_price_sum}")

elif total_price_sum >= 5000:
    print("Total price exceeds 5,000. Applying discount of 5%.")
    discount = total_price_sum * 0.5
    total_price_sum -= discount
    print(f"Discount: {discount}, New Total Price: {total_price_sum}")
```
